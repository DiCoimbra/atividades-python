#14

```python
numero = 67
tentativas = 1

while True:
  chute = int(input("Qual é o número secreto? "))
  if chute == numero:
    print(f"PARABÉNS! Você adivinhou o número corretamente em {tentativas} tentativa(s)!")
    break
  elif chute < numero:
    print("INCORRETO! Experimente um número maior")
    tentativas += 1
  else:
    print("INCORRETO! Experimente um número menor")
    tentativas += 1
```

#15

```python
positivos = 0
negativos = 0
soma = 0

while True:
  i = int(input("Digite um número: "))
  if i == 0:
    print("Número zero detectado. Programa encerrado!")
    break
  elif i > 0:
    positivos += 1
    soma += i
  else:
    negativos += 1
    soma += i

digitados = positivos + negativos

print (
  40*"—"+"\n",
  f"Números digitados: {digitados}\n",
  f"Números positivos: {positivos}\n",
  f"Números negativos: {negativos}\n",
  f"Somatório total: {soma}"
)
```
