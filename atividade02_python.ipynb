```python
s = 500
print("Operações:\n1. Sacar;\n2. Depositar;")

for i in range(0, 6):
  print(f"Saldo atual: R${s}")
  a = input("Escolha uma das duas operações (1 ou 2): ")

  if a == "1":
    v = float(input("Digite o valor do saque: "))
    while v > s:
      v = float(input(f"Saldo insuficiente! Digite um valor menor ou igual a {s}: "))

    s -= v
  elif a == "2":
    v = float(input("Digite o valor do depósito: "))
    s += v
  else:
    print("Operação não identificada!")

print(f"Saldo final: {s}")

```

```python
#facilmente o código menos otimizado e suscetível a crashar que ja fiz na vida

q = 0
l = []

a = int(input("Digite um número inteiro positivo: "))
b = int(input("Digite um número maior que o anterior: "))

while not (a < b):
  a = int(input("Digite um número inteiro: "))
  b = int(input("Digite um número maior que o anterior: "))


for i in range(a, b+1):
  primo = True

  for j in range(2, i):
    if i % j == 0:
      primo = False

  if primo == True:
    q += 1
    l.append(i)


print(f"Entre os números {a} e {b}, foram encontrados {q} números primos, sendo eles: {l}")
```

#15

```python
saldo = 200 #saldo inicial/atual/final
depositos = 0 #contador da quantidade de depositos
valor_depositos = 0 #acumulador do valor dos depositos
saques = 0 #contador da quantidade de saques
valor_saques = 0 #acumulador do valor dos saques


while True:
  print(70 * "-")
  print(f"Operações:\n0. Encerrar\n1. Depósito\n2. Saque\nSaldo: R${saldo}")
  i = input("Digite o número correspondente à operação desejada (0/1/2): ") #é a entrada que reconhece a escolha (o i é de *i*nput)

  if i == "1":
    j = float(input("Digite o valor do depósito: ")) #armazena o valor temporário do depósito/saque
    saldo += j

    depositos += 1
    valor_depositos += j
  elif i == "2":
    while True:
      j = float(input("Digite o valor do saque: "))

      if j <= saldo:
        saldo -= j

        saques += 1
        valor_saques += j
        break
      else:
        print("Saldo insuficiente!")

  elif i == "0":
    break
  else:
    print("Operação inválida")


print(70 * "-", f"\nO saldo final é R${saldo}.\nForam realizados {depositos} depositos, acumulando R${valor_depositos}, e {saques} saques, acumulando R${valor_saques}. Ao todo, foram feitas {depositos + saques} operações.")
```
