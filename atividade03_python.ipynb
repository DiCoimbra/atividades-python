```python
n = int(input("Digite um número inteiro: "))

maior = n
menor = n

p_maior = 1
p_menor = 1

l = [n]


for i in range(2, 9):
  n = int(input("Digite um número inteiro: "))

  if n > maior:
    maior = n
    if not n in l:
      p_maior = i
  elif n < menor:
    menor = n
    if not n in l:
      p_menor = i

  l.append(n)


print(20 * "-", f"\nO maior número digitado, aparecendo pela primeira vez na posição {p_maior}, foi {maior}\nO menor número digitado, aparecendo na primeira vez posição {p_menor}, foi {menor}")
```

```python
l = []
pares = 0

for i in range(0, 6):
  n = int(input("Digite um número inteiro: "))
  l.append(n)

t = tuple(l)

if 5 in t:
  a = False
  q = t.count(5)
  c = 1
  for i in t:
    if i % 2 == 0:
      pares += 1
    elif i == 5 and not a:
      a = True
      pos = c

    c += 1

  if not a:
    pos = "Não apareceu nenhuma vez"


print(20 * "-", f"\nAparições do número 5: {q}\nPrimeira aparição do número 5: Posição {pos}\nQuantidade de pares: {pares}")
```

```python
produtos = {
  "Carfentanila": 20,
  "Sufentanila": 30,
  "Fentanil": 50,
  "Etorfina": 35
}

nome = input("Digite o nome do produto: ").capitalize()

if nome in produtos.keys():
  print(f"{produtos[nome]} unidades de {nome} em estoque")

  qtd = int(input("Digite a quantidade desejada: "))

  if qtd <= produtos[nome]:
    produtos[nome] -= qtd

    print(50 * "-", f"\nAtualização do estoque:")
    for i, j in produtos.items():
      print(f"{i}: {j}")
  else:
    print("A quantidade em estoque é insuficiente!")
else:
  print("O produto não está em estoque!")
```

```python
#NOTA: eu até poderia calcular tudo o que é solicitado de forma mais enxuta e em apenas um loop,
#mas ficaria visivelmente caótico, então fiz um loop pra cada função :P

produtos = []

#CADASTRO
for i in range(0, 5):
  print(40 * "-", f"\nProduto {i + 1}")

  codigo = input("Cadastre o código: ")
  nome = input("Cadastre o nome: ")
  qtd = int(input("Cadastre a quantidade: "))
  preco = float(input("Cadastre o preço unitário: "))

  dicionario = {
    "Código": codigo,
    "Nome": nome,
    "Quantidade": qtd,
    "Preço": preco
  }

  produtos.append(dicionario)

#EXIBIÇÃO
print(70 * "-", f"\nTABELA DE PRODUTOS")
for i in produtos:
  print(50 * "-")
  for j, k in i.items():
    print(f"{j}: {k}")

#VALOR TOTAL
valor = 0
for i in produtos:
  valor += i["Quantidade"] * i["Preço"]

print(f"--> Valor total dos produtos: {valor} <--")

#ITEM DE MAIOR VALOR (DESGRAÇA mal-otimizada)
maior = produtos[0]
val_maior = maior["Quantidade"] * maior["Preço"]

for i in produtos:
  a = i["Quantidade"] * i["Preço"]
  b = maior["Quantidade"] * maior["Preço"]

  if a >= b:
    maior = i
    val_maior = a

print("--> ITEM DE MAIOR VALOR <--\n")
for i, j in maior.items():
  print(f"{i}: {j}")
print(f"Valor total: {val_maior}")

#TUPLA RESUMO
qtd = 0
val_estoque = 0
for i in produtos:
  qtd += i["Quantidade"]
  val_estoque += i["Quantidade"] * i["Preço"]

tupla_resumo = (qtd, val_estoque)
print(f"--> QUANTIDADE E VALOR DO ESTOQUE <--\nQuantidade: {tupla_resumo[0]}\nValor do Estoque: {tupla_resumo[1]}")
```
